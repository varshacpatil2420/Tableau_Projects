```python
train_name = "Rajdhani Express"
seat_number = "B2-45"
ticket_price = 2450
journey_date = "25-May-2026"
```

```python
print("Train Name   : ", train_name)
print("Seat Number  : ", seat_number)
print("Ticket Price : ",ticket_price)
print("Journey Date : ", journey_date)
```

```python
number_of_passenger = 3
```

```python
total_fare = number_of_passenger * ticket_price
print("Total Fare : ", total_fare)
```

```python
gst_percentage = 5
```

```python
gst_amount = (total_fare * gst_percentage) / 100
print("GST Amount : ", gst_amount)
```

```python
final_amount = total_fare + gst_amount
print("Total amount : ", final_amount)
```
